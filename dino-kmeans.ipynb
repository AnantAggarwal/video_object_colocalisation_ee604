```python
download_links = [
    'https://data.vision.ee.ethz.ch/cvl/youtube-objects/categories/aeroplane.tar.gz',
    'https://data.vision.ee.ethz.ch/cvl/youtube-objects/categories/bird.tar.gz',
    'https://data.vision.ee.ethz.ch/cvl/youtube-objects/categories/boat.tar.gz',
    'https://data.vision.ee.ethz.ch/cvl/youtube-objects/categories/car.tar.gz',
    'https://data.vision.ee.ethz.ch/cvl/youtube-objects/categories/cat.tar.gz',
    'https://data.vision.ee.ethz.ch/cvl/youtube-objects/categories/cow.tar.gz',
    'https://data.vision.ee.ethz.ch/cvl/youtube-objects/categories/cow.tar.gz',
    'https://data.vision.ee.ethz.ch/cvl/youtube-objects/categories/motorbike.tar.gz',
    'https://data.vision.ee.ethz.ch/cvl/youtube-objects/categories/train.tar.gz'
]

```

```python
import os
import requests
import shutil

output_directory = "yod"

os.makedirs(output_directory, exist_ok=True)
print(f"Output directory set to: {output_directory}")

for url in download_links:
    try:
        filename = url.split('/')[-1]
        save_path = os.path.join(output_directory, filename)
        
        print(f"\nDownloading: {filename}...")
        
        response = requests.get(url, stream=True)
        response.raise_for_status()  # Raise an error for bad responses (4xx or 5xx)

        with open(save_path, 'wb') as f:
            for chunk in response.iter_content(chunk_size=8192):
                f.write(chunk)
                
        print(f"Successfully downloaded to: {save_path}")
        print(f"Extracting: {filename}...")
        
        shutil.unpack_archive(save_path, output_directory)
        
        print(f"Successfully extracted: {filename}")
        os.remove(save_path)
        print(f"Removed archive: {filename}")

    except requests.exceptions.RequestException as e:
        print(f"Error downloading {url}: {e}")
    except Exception as e:
        print(f"Error processing {url}: {e}")

print("\n--- All operations complete! ---")
```
