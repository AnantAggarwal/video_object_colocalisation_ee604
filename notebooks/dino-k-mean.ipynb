```python
import torch
import cv2
import numpy as np
import os
from transformers import AutoImageProcessor, AutoModel
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
import joblib
from tqdm import tqdm
import matplotlib.pyplot as plt
```

```python
OUTPUT_DIR = 'dinov2_coloc_output'
PCA_MODEL_PATH = os.path.join(OUTPUT_DIR, 'pca_model.joblib')
KMEANS_MODEL_PATH = os.path.join(OUTPUT_DIR, 'kmeans_model.joblib')
VIS_FILE_PATH = os.path.join(OUTPUT_DIR, 'cluster_visualization.jpg')

MODEL_NAME = 'facebook/dinov2-small'
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

PCA_COMPONENTS = 128    # PCA dimension (DINOv2 'small' is 384D)
N_CLUSTERS = 16
KMEANS_SUBSET_RATIO = 0.1
```

```python
def get_frames(frame_paths):
    frames = []
    for path in tqdm(frame_paths, total=len(frame_paths)):
        img = cv2.imread(path)
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        frames.append(img)

    return frames

#frames = get_frames(frame_paths)
```

```python
@torch.no_grad()
def extract_features(frames, model, processor):
    """Extracts DINOv2 patch features for a batch of frames."""
    if not frames:
        return None
    inputs = processor(images=frames, return_tensors="pt", do_resize=True, size=224).to(DEVICE)
    outputs = model(**inputs)
    patch_features = outputs.last_hidden_state[:, 1:, :]
    num_frames, num_patches, dim = patch_features.shape
    return patch_features.reshape(num_frames * num_patches, dim).cpu()
```

```python
PATCH_GRID = (16,16)
```

```python
def visualize_clusters(frame, patch_labels, n_clusters, patch_grid_size=(16, 16)):
    """Creates a visualization of all K-Means clusters for a single frame."""
    
    # Resize frame to match the processor's input size for accurate overlay
    frame_resized = cv2.resize(frame, (224, 224))

    # Create the full segmentation mask
    mask = patch_labels.reshape(patch_grid_size)
    mask_full = cv2.resize(mask.astype(np.uint8), (224, 224), interpolation=cv2.INTER_NEAREST)

    # Create a composite image
    fig, axes = plt.subplots(1, n_clusters + 1, figsize=(20, 4))
    
    # Original Image
    axes[0].imshow(frame_resized)
    axes[0].set_title("Original")
    axes[0].axis('off')

    # Each cluster mask
    for i in range(n_clusters):
        binary_mask = (mask_full == i).astype(np.uint8)
        
        # Create a red overlay
        overlay = np.zeros_like(frame_resized, dtype=np.uint8)
        overlay[..., 0] = 150  # Red
        
        # Apply overlay where mask is active
        masked_img = cv2.bitwise_and(overlay, overlay, mask=binary_mask)
        
        # Blend with original
        display_img = cv2.addWeighted(frame_resized, 1.0, masked_img, 0.7, 0)
        
        axes[i+1].imshow(display_img)
        axes[i+1].set_title(f"Cluster {i}")
        axes[i+1].axis('off')
    
    plt.tight_layout()
    return fig
```

```python
os.makedirs(OUTPUT_DIR, exist_ok=True)

print(f"Using device: {DEVICE}")
print("Loading DINOv2 model and processor...")
processor = AutoImageProcessor.from_pretrained(MODEL_NAME)
model = AutoModel.from_pretrained(MODEL_NAME).to(DEVICE).eval()

patch_size = model.config.patch_size
print(processor.size)
grid_size = 224 // patch_size
PATCH_GRID = (grid_size, grid_size)

print("\n--- Part 1: Extracting Features ---")
all_features_list = []

for class_name in ['car']:
    print(f"\nProcessing class: {class_name}")
    frames = get_frames(frame_paths[::5])
    if not frames:
        continue
    
    # Batch processing for efficiency (adjust batch_size based on VRAM)
    batch_size = 32
    for i in tqdm(range(0, len(frames), batch_size), desc=f"Extracting {class_name}"):
        batch_frames = frames[i:i+batch_size]
        features = extract_features(batch_frames, model, processor)
        if features is not None:
            all_features_list.append(features)
```

```python
all_features = torch.cat(all_features_list, dim=0).numpy()
print(f"\nTotal features extracted: {all_features.shape}")
```

```python
print("\n--- Part 2: Running PCA and K-Means ---")

print(f"Running PCA to reduce to {PCA_COMPONENTS} dims...")
pca = PCA(n_components=PCA_COMPONENTS)
features_pca = pca.fit_transform(all_features)

print(f"Training K-Means on a {KMEANS_SUBSET_RATIO*100}% subset of features...")
# Create a random subset for fast K-Means training
n_samples = features_pca.shape[0]
subset_indices = np.random.choice(n_samples, int(n_samples * KMEANS_SUBSET_RATIO), replace=False)
features_subset = features_pca[subset_indices]

kmeans = KMeans(n_clusters=N_CLUSTERS, random_state=42, n_init=10)
kmeans.fit(features_subset)

print("Saving PCA and K-Means models...")
joblib.dump(pca, PCA_MODEL_PATH)
joblib.dump(kmeans, KMEANS_MODEL_PATH)
```

```python
import gradio as gr
import torch
import cv2
import numpy as np
import os
import joblib
from transformers import AutoImageProcessor, AutoModel
from sklearn.decomposition import PCA
from sklearn.preprocessing import MinMaxScaler
from tqdm import tqdm
import warnings

# --- 1. Configuration & Model Paths ---

# !!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!
# !! USER: YOU MUST EDIT THESE PATHS
# !!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!
PCA_MODEL_PATH = 'dinov2_coloc_output/pca_model.joblib'
KMEANS_MODEL_PATH = 'dinov2_coloc_output/kmeans_model.joblib'
# !!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!!

# --- Model Parameters ---
MODEL_NAME = 'facebook/dinov2-small'
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# --- Cluster & Feature Parameters (from your training) ---
TARGET_IMAGE_SIZE = 224
PATCH_SIZE = 14
GRID_SIZE = TARGET_IMAGE_SIZE // PATCH_SIZE
PATCH_GRID = (GRID_SIZE, GRID_SIZE)
PATCHES_PER_FRAME = PATCH_GRID[0] * PATCH_GRID[1]

# Suppress a specific warning from scikit-learn/joblib
warnings.filterwarnings("ignore", category=UserWarning, message="Trying to unpickle estimator")

# --- 2. Load All Models (Called ONCE at startup) ---

def load_models():
    """Loads all models and generates the cluster color map."""
    print("Loading models... This may take a moment.")
    
    if not os.path.exists(PCA_MODEL_PATH):
        raise FileNotFoundError(f"PCA model not found at: {PCA_MODEL_PATH}")
    if not os.path.exists(KMEANS_MODEL_PATH):
        raise FileNotFoundError(f"K-Means model not found at: {KMEANS_MODEL_PATH}")
        
    models = {}
    try:
        models['processor'] = AutoImageProcessor.from_pretrained(MODEL_NAME)
        models['dino'] = AutoModel.from_pretrained(MODEL_NAME).to(DEVICE).eval()
        models['pca'] = joblib.load(PCA_MODEL_PATH)
        models['kmeans'] = joblib.load(KMEANS_MODEL_PATH)
        
        # --- Generate the Cluster Color Map ---
        print("Generating 3D color map from K-Means centroids...")
        centroids_pca = models['kmeans'].cluster_centers_
        
        # Use a new PCA to project centroids from 64D down to 3D
        pca_color = PCA(n_components=3)
        centroids_3d = pca_color.fit_transform(centroids_pca)
        
        # Normalize to 0.0-1.0 to create RGB colors
        scaler = MinMaxScaler()
        color_map = scaler.fit_transform(centroids_3d)
        
        # Store the color map for use
        models['color_map'] = color_map
        print(f"Color map generated for {len(color_map)} clusters.")
            
    except Exception as e:
        print(f"Error loading models: {e}")
        raise
        
    print("Models loaded successfully.")
    return models

MODELS = load_models()


# --- 3. Core Helper Functions ---

@torch.no_grad()
def extract_features(frame_rgb, model, processor):
    """Extracts DINOv2 patch features for a single frame."""
    inputs = processor(
        images=[frame_rgb], # Note: expects a list
        return_tensors="pt",
        do_resize=True,
        size={'height': TARGET_IMAGE_SIZE, 'width': TARGET_IMAGE_SIZE}
    ).to(DEVICE)
    
    outputs = model(**inputs)
    patch_features = outputs.last_hidden_state[:, 1:, :] # (1, num_patches, dim)
    
    return patch_features.reshape(PATCHES_PER_FRAME, -1).cpu().numpy()

def create_multi_color_overlay(frame, patch_labels, color_map_bgr, frame_wh):
    """
    Creates a full multi-color blended overlay for all clusters.
    """
    width, height = frame_wh
    
    # Create the low-res patch map
    mask = patch_labels.reshape(PATCH_GRID)
    mask_rgb_lowres = np.zeros((*PATCH_GRID, 3), dtype=np.float32)
    
    # "Paint" the low-res mask with cluster colors
    for k in range(len(color_map_bgr)):
        mask_rgb_lowres[mask == k] = color_map_bgr[k]
        
    # Upscale the color mask to the full frame size
    mask_rgb_full = cv2.resize(mask_rgb_lowres, (width, height), interpolation=cv2.INTER_NEAREST)
    
    # Convert frame to float for blending
    frame_float = frame.astype(np.float32) / 255.0
    
    # Blend the frame with the mask
    blended_img = cv2.addWeighted(frame_float, 0.6, mask_rgb_full, 0.4, 0)
    
    # Convert back to uint8
    blended_img_uint8 = (blended_img * 255).astype(np.uint8)
    return blended_img_uint8

def process_video_all_clusters(input_video_path, progress_callback):
    """
    Processes a single video, highlights ALL clusters, and returns the output path.
    """
    cap = cv2.VideoCapture(input_video_path)
    if not cap.isOpened():
        raise gr.Error(f"Could not open video file: {input_video_path}")

    # Get video properties
    fps = cap.get(cv2.CAP_PROP_FPS)
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    
    # Create a temporary output file
    output_video_path = f"{input_video_path}.all_clusters.mp4"
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(output_video_path, fourcc, fps, (width, height))

    # Load models from the global dict
    processor = MODELS['processor']
    dino = MODELS['dino']
    pca = MODELS['pca']
    kmeans = MODELS['kmeans']
    
    # Convert RGB color map (for float) to BGR (for OpenCV uint8)
    color_map_rgb = MODELS['color_map']
    color_map_bgr = color_map_rgb[..., ::-1] # Flip R and B channels

    for i in tqdm(range(frame_count), desc=f"Processing Video (All Clusters)"):
        ret, frame = cap.read()
        if not ret:
            break
        
        # 1. Convert to RGB
        frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        
        # 2. Extract Features
        features = extract_features(frame_rgb, dino, processor)
        
        # 3. PCA + K-Means
        features_pca = pca.transform(features)
        patch_labels = kmeans.predict(features_pca) # Shape (256,)
        
        # 4. Create Multi-Color Highlighted Frame
        highlighted_frame = create_multi_color_overlay(
            frame,
            patch_labels,
            color_map_bgr,
            (width, height)
        )
        
        # 5. Write to output video
        out.write(highlighted_frame)
        
        # 6. Update Gradio progress
        progress_callback((i + 1) / frame_count)

    cap.release()
    out.release()
    
    return output_video_path

# --- 4. Gradio Main Function ---

def highlight_all_clusters(video1_path, video2_path, progress=gr.Progress()):
    """
    This is the main function called by the Gradio button.
    It orchestrates the processing of both videos.
    """
    if video1_path is None or video2_path is None:
        raise gr.Error("Please upload both videos.")
        
    print("Starting job. Highlighting ALL Clusters.")

    # Process Video 1
    progress(0, desc="Starting Video 1...")
    out1_path = process_video_all_clusters(
        video1_path,
        progress_callback=lambda p: progress(p * 0.5, desc="Processing Video 1...")
    )
    
    # Process Video 2
    progress(0.5, desc="Starting Video 2...")
    out2_path = process_video_all_clusters(
        video2_path,
        progress_callback=lambda p: progress(0.5 + p * 0.5, desc="Processing Video 2...")
    )

    print("Job complete.")
    return out1_path, out2_path

# --- 5. Build Gradio App ---

with gr.Blocks(theme=gr.themes.Soft()) as demo:
    gr.Markdown(
        """
        # DINOv2 Unsupervised Co-Localization Demo
        Upload two videos that contain a common object (e.g., two different videos of cars).
        
        1.  Click **"Generate Cluster Maps"**.
        2.  The app will process both videos and generate a "cluster map" for every frame.
        3.  Each cluster is assigned a unique color. Clusters that are "closer" in the feature space
            will have similar colors.
        
        **Your Goal:** Watch both output videos and see if the *same color* (e.g., "blue")
        consistently highlights the common object (e.g., the "car") in both.
        """
    )
    
    with gr.Row():
        video_in1 = gr.Video(label="Input Video 1")
        video_in2 = gr.Video(label="Input Video 2")

    btn = gr.Button("Generate Cluster Maps")
    
    with gr.Row():
        video_out1 = gr.Video(label="Output Video 1 (Cluster Map)")
        video_out2 = gr.Video(label="Output Video 2 (Cluster Map)")

    btn.click(
        fn=highlight_all_clusters,
        inputs=[video_in1, video_in2],
        outputs=[video_out1, video_out2]
    )

if __name__ == "__main__":
    demo.launch(debug=True)
```
